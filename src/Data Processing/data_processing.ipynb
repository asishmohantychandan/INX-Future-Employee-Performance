### 1.8 Variable-Level Data Understanding

Each variable is examined through its data type, number of unique values, minimum and maximum values where applicable, and representative categories.

This assessment helps distinguish continuous numerical variables, ordinal variables, nominal categorical variables, identifier variables, and the target variable before preprocessing.

```python
# Create a compact variable-level summary

variable_summary = []

for column in df.columns:
    variable_info = {
        "Column": column,
        "Data Type": str(df[column].dtype),
        "Unique Values": df[column].nunique(),
        "Missing Values": df[column].isna().sum()
    }

    if pd.api.types.is_numeric_dtype(df[column]):
        variable_info["Minimum"] = df[column].min()
        variable_info["Maximum"] = df[column].max()
        variable_info["Example Values"] = sorted(
            df[column].unique()
        )[:10]
    else:
        variable_info["Minimum"] = "-"
        variable_info["Maximum"] = "-"
        variable_info["Example Values"] = df[column].unique()[:10].tolist()

    variable_summary.append(variable_info)

variable_summary_df = pd.DataFrame(variable_summary)

variable_summary_df
```

### 1.10 Logical Consistency Checks

Logical consistency checks are performed on employee experience-related variables.

The purpose is to identify records where the relationship between different experience measures appears inconsistent, such as experience in the current role exceeding total experience at the company.

```python
# Logical consistency checks for experience-related variables

check_current_role = (
    df["ExperienceYearsInCurrentRole"]
    > df["ExperienceYearsAtThisCompany"]
)

check_company_experience = (
    df["ExperienceYearsAtThisCompany"]
    > df["TotalWorkExperienceInYears"]
)

check_manager_experience = (
    df["YearsWithCurrManager"]
    > df["ExperienceYearsAtThisCompany"]
)

print(
    "Current role experience > company experience:",
    check_current_role.sum()
)

print(
    "Company experience > total work experience:",
    check_company_experience.sum()
)

print(
    "Manager experience > company experience:",
    check_manager_experience.sum()
)
```

### 1.13 Feature Classification

The variables are classified according to their role and measurement type before feature engineering.

The classification distinguishes:

- Target variable
- Identifier variable
- Ordinal variables, where the numerical levels represent an ordered scale
- Nominal categorical variables, where categories do not have a natural numerical order
- Numerical variables, representing measurable quantities or counts

This classification is used to determine the appropriate preprocessing approach during model development.

```python
# Target variable
target_column = "PerformanceRating"

# Identifier variable
identifier_columns = [
    "EmpNumber"
]

# Ordinal variables
ordinal_columns = [
    "EmpEducationLevel",
    "EmpEnvironmentSatisfaction",
    "EmpJobInvolvement",
    "EmpJobSatisfaction",
    "EmpRelationshipSatisfaction",
    "EmpWorkLifeBalance"
]

# Nominal categorical variables
categorical_columns = [
    "Gender",
    "EducationBackground",
    "MaritalStatus",
    "EmpDepartment",
    "EmpJobRole",
    "BusinessTravelFrequency",
    "OverTime",
    "Attrition"
]

# Numerical variables
numerical_columns = [
    "Age",
    "DistanceFromHome",
    "EmpHourlyRate",
    "EmpJobLevel",
    "NumCompaniesWorked",
    "EmpLastSalaryHikePercent",
    "TotalWorkExperienceInYears",
    "TrainingTimesLastYear",
    "ExperienceYearsAtThisCompany",
    "ExperienceYearsInCurrentRole",
    "YearsSinceLastPromotion",
    "YearsWithCurrManager"
]

# Display classification
feature_classification = pd.DataFrame({
    "Feature_Type": (
        ["Target"] +
        ["Identifier"] * len(identifier_columns) +
        ["Ordinal"] * len(ordinal_columns) +
        ["Nominal Categorical"] * len(categorical_columns) +
        ["Numerical"] * len(numerical_columns)
    ),
    "Feature": (
        [target_column] +
        identifier_columns +
        ordinal_columns +
        categorical_columns +
        numerical_columns
    )
})

display(feature_classification)

print("Total classified columns:", len(feature_classification))
print("Original dataset columns :", len(df_clean.columns))
```

### 1.14 Feature Engineering and Transformation

Feature engineering and transformation decisions are based on the variable classification and the data definitions provided with the dataset.

The ordinal variables are retained in their existing ordered numerical representation because their levels have a meaningful order.

Nominal categorical variables are retained in categorical form at this stage. They will be encoded during model development using a preprocessing pipeline.

Numerical variables are retained in their original representation. Scaling will be applied within the model-specific preprocessing pipeline where required.

The employee identifier `EmpNumber` is excluded from the modeling features because it is an identifier rather than an employee characteristic.

The target variable `PerformanceRating` is kept separately from the predictor variables.

No artificial features are created at this stage because the available variables already represent employee characteristics, job-related factors, satisfaction measures, experience measures, and organizational attributes required for the analysis.

```python
# Create the modeling feature set by excluding the identifier and target
X = df_clean.drop(
    columns=[target_column] + identifier_columns
)

y = df_clean[target_column].copy()

print("Predictor dataset shape:", X.shape)
print("Target shape           :", y.shape)

print("\nPredictor columns:")
print(X.columns.tolist())

print("\nTarget distribution:")
print(y.value_counts().sort_index())
```

### 1.15 Prepare Modeling Dataset

The final modeling dataset is prepared from the cleaned employee data.

The employee identifier is excluded because it does not represent a predictive employee characteristic. The target variable is kept separately from the predictor variables.

The processed dataset retains the original feature representations at this stage. Encoding of nominal categorical variables and scaling of numerical variables will be performed later within the model-training preprocessing pipelines.

The final dataset is validated before being saved to the processed-data directory.

```python
# Create the final modeling dataset
modeling_data = X.copy()

# Add the target variable
modeling_data[target_column] = y

print("Final modeling dataset shape:", modeling_data.shape)

print("\nMissing values:")
print(modeling_data.isnull().sum().sum())

print("\nDuplicate rows:")
print(modeling_data.duplicated().sum())

print("\nTarget distribution:")
print(modeling_data[target_column].value_counts().sort_index())

print("\nData types:")
display(modeling_data.dtypes.to_frame(name="Data Type"))
```

### 1.16 Save Processed Dataset

The validated modeling dataset is saved in the processed-data directory.

The raw dataset remains unchanged in the raw-data directory. The processed dataset serves as the canonical input for exploratory analysis and model development.

```python
import os

# Path for the processed dataset
processed_path = "../../data/processed/employee_performance_processed.csv"

# Create the processed directory if it does not already exist
os.makedirs(os.path.dirname(processed_path), exist_ok=True)

# Save the processed dataset
modeling_data.to_csv(
    processed_path,
    index=False
)

print("Processed dataset saved successfully.")
print("Path:", processed_path)
print("Shape:", modeling_data.shape)
```
